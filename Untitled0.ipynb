```python
import pandas as pd
import librosa
import numpy as np
import os

# Define file paths (Adjust these to your local paths)
AUDIO_DIR = './audios/'
TRAIN_CSV = 'train.csv'
TEST_CSV = 'test.csv'

# Load the training data
df_train = pd.read_csv(TRAIN_CSV)
# df_test = pd.read_csv(TEST_CSV) # Load this later for prediction

# The label is likely a continuous score (e.g., 1.0 to 5.0) or a categorical one.
# For demonstration, we assume it's a regression task (a score).
print(f"Training data shape: {df_train.shape}")
print(df_train.head())

# Function to load audio file and calculate basic features (optional, but good for diagnostics)
def get_audio_features(file_name):
    path = os.path.join(AUDIO_DIR, file_name)
    try:
        y, sr = librosa.load(path, sr=None) # y: audio time series, sr: sampling rate

        # Example of a simple feature: duration
        duration = librosa.get_duration(y=y, sr=sr)

        return {'duration': duration}
    except Exception as e:
        print(f"Error loading {file_name}: {e}")
        return {'duration': np.nan}

# Example of applying the function (this can take a long time!)
# df_train['features'] = df_train['file_name'].apply(get_audio_features)
```

```python
from transformers import pipeline
import torch

# Initialize the ASR pipeline with the Whisper model
# 'medium' or 'large' will give better accuracy but is very resource-intensive (GPU recommended)
# For quick testing, you can try 'small' or 'base'.
asr_pipeline = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-large-v2", # Best model for accuracy
    device=0 if torch.cuda.is_available() else -1 # Use GPU if available
)

# Function to transcribe a single audio file
def transcribe_audio(file_name):
    path = os.path.join(AUDIO_DIR, file_name)
    try:
        # The pipeline handles loading the audio
        result = asr_pipeline(path)
        return result['text']
    except Exception as e:
        print(f"ASR Error on {file_name}: {e}")
        return ""

# Apply transcription to the training data (THIS WILL TAKE HOURS on a CPU)
# You should save the results to a new CSV file immediately after running this!
# df_train['transcribed_text'] = df_train['file_name'].apply(transcribe_audio)

# --- ASSUMPTION: The transcription step is complete and saved to a file ---
# Load the data with the transcribed text
# df_train_transcribed = pd.read_csv('train_with_transcriptions.csv')
```

```python
import spacy
from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer

# Load the spaCy English model for linguistic analysis
try:
    nlp = spacy.load("en_core_web_sm")
except:
    # If the model is not downloaded, run: python -m spacy download en_core_web_sm
    print("SpaCy model not found. Please run: python -m spacy download en_core_web_sm")
    # You will need to install and download spaCy: pip install spacy && python -m spacy download en_core_web_sm

def extract_nlp_features(text):
    if not text:
        return {}

    doc = nlp(text.lower())

    # Feature 1: Word Count
    word_count = len([token for token in doc if token.is_alpha])

    # Feature 2: POS Distribution
    pos_counts = Counter([token.pos_ for token in doc])

    # Feature 3: Lexical Diversity (Type-Token Ratio)
    unique_words = set([token.text for token in doc if token.is_alpha])
    ttr = len(unique_words) / word_count if word_count > 0 else 0

    # Feature 4: Average Sentence Length
    sentences = list(doc.sents)
    avg_sentence_len = sum(len(sent) for sent in sentences) / len(sentences) if len(sentences) > 0 else 0

    features = {
        'word_count': word_count,
        'ttr': ttr,
        'avg_sentence_len': avg_sentence_len,
        **{f'POS_{pos}': pos_counts.get(pos, 0) for pos in ['NOUN', 'VERB', 'ADJ', 'ADV', 'PRON', 'DET', 'ADP', 'AUX']}
    }

    return features

# --- Example of creating the feature matrix ---
# Assuming 'df_train_transcribed' has the 'transcribed_text' column
# features_list = [extract_nlp_features(text) for text in df_train_transcribed['transcribed_text']]
# df_features = pd.DataFrame(features_list)
# X = df_features.values
# y = df_train_transcribed['grammar_score_label'].values # Use your actual label column
```
